```python
def print_list(lst):
  for items in lst:
    print(items, end=" ")

numbers = [1,2,3,4,5]
print_list(numbers)
```

```python
def factorial(n):

  fact = 1

  for i in range(1, n+1):
    fact *= i
  return fact
print(factorial(6))
```

```python
def usd_to_inr():
  usd_amount = int(input("Enter the amount in USD: "))
  inr = usd_amount * 94
  print("INR =", inr)

usd_to_inr()
```
